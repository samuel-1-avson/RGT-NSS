# Week 3: Data Analysis and Visualization

This notebook analyzes the cleaned Superstore dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Load cleaned data
df = pd.read_csv('../data/superstore_cleaned.csv', parse_dates=['Order Date', 'Ship Date'])
print(f"Loaded {len(df):,} rows of cleaned data")
```

## Analysis 1: Sales by Category

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

print("Sales by Category:")
print("-"*40)
for cat, sales in category_sales.items():
    print(f"{cat}: ${sales:,.2f}")

# Visualization
plt.figure(figsize=(10, 6))
category_sales.plot(kind='bar', color='steelblue')
plt.title('Total Sales by Category', fontsize=14)
plt.ylabel('Sales ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('../outputs/sales_by_category.png', dpi=300)
plt.show()
```

## Analysis 2: Profit by Region

```python
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)

print("\nProfit by Region:")
print("-"*40)
for region, profit in region_profit.items():
    print(f"{region}: ${profit:,.2f}")

# Visualization
plt.figure(figsize=(10, 6))
colors = ['green' if x > 0 else 'red' for x in region_profit.values]
region_profit.plot(kind='bar', color=colors)
plt.title('Total Profit by Region', fontsize=14)
plt.ylabel('Profit ($)')
plt.axhline(y=0, color='black', linestyle='-', linewidth=0.5)
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('../outputs/profit_by_region.png', dpi=300)
plt.show()
```

## Analysis 3: Monthly Sales Trend

```python
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()

# Visualization
plt.figure(figsize=(14, 6))
monthly_sales.plot(kind='line', marker='o', color='steelblue')
plt.title('Monthly Sales Trend', fontsize=14)
plt.ylabel('Sales ($)')
plt.xlabel('Month')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../outputs/monthly_sales_trend.png', dpi=300)
plt.show()

print(f"\nTotal months: {len(monthly_sales)}")
print(f"Average monthly sales: ${monthly_sales.mean():,.2f}")
```

## Analysis 4: Profit vs Sales Scatter

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Sales'], df['Profit'], alpha=0.5, c='steelblue')
plt.xlabel('Sales ($)')
plt.ylabel('Profit ($)')
plt.title('Profit vs Sales Relationship', fontsize=14)
plt.axhline(y=0, color='red', linestyle='--', alpha=0.5, label='Break-even')
plt.legend()
plt.tight_layout()
plt.savefig('../outputs/profit_vs_sales.png', dpi=300)
plt.show()

# Calculate correlation
correlation = df['Sales'].corr(df['Profit'])
print(f"\nSales-Profit Correlation: {correlation:.3f}")
```

## Analysis 5: Top Sub-Categories

```python
subcat_profit = df.groupby('Sub-Category').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).sort_values('Profit', ascending=False)

print("\nTop 10 Sub-Categories by Profit:")
print("-"*50)
print(subcat_profit.head(10).round(2))

# Visualization
plt.figure(figsize=(12, 6))
subcat_profit['Profit'].head(10).plot(kind='barh', color='steelblue')
plt.title('Top 10 Sub-Categories by Profit', fontsize=14)
plt.xlabel('Profit ($)')
plt.tight_layout()
plt.savefig('../outputs/top_subcategories.png', dpi=300)
plt.show()
```

## Analysis 6: Shipping Performance

```python
if 'Shipping Days' in df.columns:
    print("\nShipping Performance:")
    print("-"*40)
    print(df['Shipping Days'].describe().round(2))
    
    # By Ship Mode
    shipping_by_mode = df.groupby('Ship Mode')['Shipping Days'].mean().sort_values()
    
    plt.figure(figsize=(10, 6))
    shipping_by_mode.plot(kind='bar', color='steelblue')
    plt.title('Average Shipping Days by Ship Mode', fontsize=14)
    plt.ylabel('Days')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('../outputs/shipping_by_mode.png', dpi=300)
    plt.show()
```

## Key Insights Summary

1. **Technology** leads in sales, followed by Furniture and Office Supplies
2. **West Region** generates the most profit, while some regions show losses
3. **Sales trend** shows seasonal patterns with peaks in certain months
4. **Profit margin** varies significantly across sub-categories
5. **Shipping mode** significantly impacts delivery time
