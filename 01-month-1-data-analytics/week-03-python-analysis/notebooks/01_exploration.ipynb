# Week 3: Data Loading and Exploration

**Dataset**: Superstore Sales

This notebook covers initial exploration of the Superstore dataset before cleaning.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

print("✓ Libraries imported")
```

## Load Data

```python
# Load Superstore data
df = pd.read_csv('../data/superstore.csv')

print("="*60)
print("DATASET OVERVIEW")
print("="*60)
print(f"\nShape: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

## Basic Information

```python
# Column information
print("COLUMNS:")
print("-"*60)
for i, col in enumerate(df.columns, 1):
    print(f"{i:2d}. {col}")

# Data types
print("\n" + "="*60)
print("DATA TYPES")
print("="*60)
print(df.dtypes)
```

## Missing Values Check

```python
# Check missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
}).sort_values('Missing Count', ascending=False)

print("="*60)
print("MISSING VALUES")
print("="*60)
print(missing_df[missing_df['Missing Count'] > 0])

if missing.sum() == 0:
    print("\n✓ No missing values found!")
else:
    print(f"\n⚠ Found {missing.sum()} missing values")
```

## Duplicate Check

```python
# Check for duplicates
duplicates = df.duplicated().sum()
print(f"Duplicate rows: {duplicates}")

if duplicates == 0:
    print("\n✓ No duplicate rows found!")
else:
    print(f"\n⚠ Found {duplicates} duplicate rows")
```

## Descriptive Statistics

```python
# Numeric columns
print("="*60)
print("NUMERIC COLUMNS")
print("="*60)
numeric_cols = df.select_dtypes(include=[np.number]).columns
print(df[numeric_cols].describe().round(2))
```

## Categorical Overview

```python
# Key categorical columns
cat_cols = ['Category', 'Sub-Category', 'Region', 'Segment', 'Ship Mode']

for col in cat_cols:
    if col in df.columns:
        print("\n" + "="*60)
        print(f"{col.upper()}")
        print("="*60)
        print(df[col].value_counts().head(10))
```

## Data Quality Summary

```python
# Summary report
print("="*60)
print("DATA QUALITY SUMMARY")
print("="*60)
print(f"\nTotal Rows: {len(df):,}")
print(f"Total Columns: {len(df.columns)}")
print(f"Duplicate Rows: {duplicates}")
print(f"Missing Values: {missing.sum()}")

print("\nIssues to Address:")
print("  1. Convert Order Date and Ship Date to datetime")
print("  2. Fill missing Postal Codes")
print("  3. Create derived features (Shipping Days, Profit Margin)")
print("  4. Check for negative values in Sales/Profit")
```
